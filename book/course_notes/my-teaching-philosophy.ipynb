I believe strong teaching should combine mathematical rigor, intuition, and practical coding fluency.

## Core Principles

1. Build intuition first, then formalize with mathematics.
2. Use coding demos to connect equations to behavior.
3. Encourage questions and iterative learning.
4. Emphasize clarity, reproducibility, and responsible AI practice.

## In Practice

- I structure lectures around a concept -> derivation -> implementation pattern.
- I prioritize live examples and visual explanations.
- I design material for both foundational understanding and research readiness.
