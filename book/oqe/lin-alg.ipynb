# The Essence of Linear Algebra

The fundamental component of Linear Algebra is a `vector`.

A mathematician's perspective of a vector is the interplay (generalization) between vectors as "pointed arrows in space" and "list of numbers" (with 1 or more dimensions, the length of list determines the dimension of a vector). In the "list of numbers" perspective, *the order matters*.

An informal definition of a vector can thought of `anything where there is a sensible notion of adding two vectors and multiplying a vector by a number`.

```{figure} figures/2d-vector.png
---
width: 50%
name: vector-simple
---

When the word `vector` is used, think of an arrow in a 2-D coordinate system (with its tail at the origin).

```

The coordinates of a vector tell about how to reach that vector from its tail (at origin). `Every pair of numbers (2D vector) gives you one and only one vector.`

The same idea can be extended to any number of dimensions, we want.

```{figure} figures/3d-vector.png
---
width: 50%
name: vector-3d
---

3-Dimensional vector

```

## Summation of two vectors

When we want to add two vectors ($\vec{u}, \vec{v}$), we shift the second vector such that the tail of the second vector is at the tip of the first vector $\vec{u}$ and the new vector from the tail of $\vec{u}$ to the tip of $\vec{v}$ is the sum of two vectors ($\vec{u}, \vec{v}$).

```{figure} figures/sum-of-two-vectors.png
---
width: 50%
name: vector-3d
---

Sum of two vectors

```

We can think of this summation as an extension to `summation along real number line`.

```{figure} figures/sum-of-two-vectors-2.png
---
width: 50%
name: vector-3d
---

Sum of two vectors - 2 

```

Multiplication of a vector by a number (scalar) is simply scaling the vector by that scalar.

Concepts in Linear Algebra revolves around the fundamental operations of `vector addition and scalar multiplication`.

## Linear Combinations, Spans, and Basis Vectors

In linear algebra we have basis vectors denoted by $\hat{i}$ and $\hat{j}$ that are unit vectors along $x$ and $y$ axes respectively.

Every vector in the 2D space is a linear combination of these `basis vectors`.

```{figure} figures/basis-vectors.png
---
width: 50%
name: vector-3d
---

Basis Vectors

```

We could have chosen different vectors as the basis vectors and could have represented all vectors that are a linear combination of these basis vectors.

`We can reach every possible 2D vectors using the basis vectors.`.

Every vector can be thought as a linear combination of the basis vectors. (In the default case $\hat{i}$ and $\hat{j}$ are perpendicular to each other and can reach every point on the real line).

```{figure} figures/linear-combination.png
---
width: 80%
name: linear-combination
---

Linear Combination

```

If we consider a general case, when two vectors $\vec{u}$ and $\vec{v}$ are considered, there can be a finite number of linear combinations with these vectors and this is known as the `span of vectors` $\vec{u}$ `and` $\vec{v}$.


```{figure} figures/span-vectors.png
---
width: 80%
name: span-vectors
---

Span of two vectors $\vec{u}$ and $\vec{v}$

```

(Span of two vectors that have $0^{\circ}$ between them is a line).

Think of single vector as a pointed arrow, but when a collection of vectors is considered, think of them as points in space. A span of two vectors can fill up the entire 2-D plane.

`Span of two vectors in a 3D plane is a plane (when the vectors are at an angle).`

Linearly dependent - angle between vectors is zero and the span fills up a line even two vectors are provided in a 2D space.

Linearly independent vectors.

## Linear Transformations and Matrices

Transformation is simply a function that takes an input and gives an output

Linear transformations limits to only simple linear functions to perform transformation.

`To understand this, think of vectors are points in space, and the transformation output moves these points in space to give the output (example: pixels are transformed from one location to another in an image after transformation).`

All we need to do is keep track of the basis vector and everything else follows naturally.


```{figure} figures/linear-transformation.png
---
width: 80%
name: linear-transformation
---

Linear Transformation of vectors.

```
